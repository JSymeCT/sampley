# ```sampley``` exemplar: additional functions

## ```generate_dfls``` - generate distances from a line

### Setup

To use this function, we have to import it from the ```functions``` module.
<br>Note that ```from sampley.functions import *``` will also import it (along with all other functions).

````python
from sampley.functions import generate_dfls
````

We also import some additional packages for the sake of this exemplar.

````python
import matplotlib.pyplot as plt
from math import exp
````

### Example 1: without detection function

First, we set the parameters.

````python
number = 100000
esw = 2000
interval = 1
dfunc = None
````

Then we run the function to get our list of distances: ```dfls```.

````python
dfls = generate_dfls(
    number=number, 
    esw=esw, 
    interval=interval, 
    dfunc=dfunc)
````

To see the result, we can print the first 10 distances and check the total number generated.

````python
print(f'First 10 distances: {dfls[:10]}\nNumber of distances: {len(dfls)}')
````

To get a better idea of what they are like, we can plot a histogram of the distances and see that they are evenly spread from 0 to the ESW.

````python
plt.hist(dfls, bins=50, facecolor='none', edgecolor='black');
````

### Example 2: with detection function

First, we set the parameters.

````python
number = 100000
esw = 2000
interval = 1
def dfunc(x): return exp((-x**2) / (2*500**2))
````

Then we run the function to get our list of distances: ```dfls```.

````python
dfls = generate_dfls(
    number=number, 
    esw=esw, 
    interval=interval, 
    dfunc=dfunc)
````

To see the result, we can print the first 10 distances and check the total number generated.

````python
print(f'First 10 distances: {dfls[:10]}\nNumber of distances: {len(dfls)}')
````

To get a better idea of what they are like, we can plot a histogram of the distances and see that they follow the detection function, with distances close to 0 being more frequent than those close to the ESW.

````python
plt.hist(dfls, bins=50, facecolor='none', edgecolor='black');
````
